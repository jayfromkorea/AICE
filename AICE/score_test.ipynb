````python
names = [make_korean_name() for _ in range(100)]
print(names)
````

````python
di = {
    '번호' : np.arange(1, 101), # 1 - 100까지 순차적인 값들의 배열 생성
    '이름' : names,
    '학년' : 3,
    '반' : np.random.randint(1, 6, size=100),   # 1 - 5까지 100개의 데이터 생성
    '국어' : np.clip(np.round(np.random.normal(40, 20, size=100), 2), 0, 100),    # 정규분포임
    '수학' : np.clip(np.round(np.random.normal(60, 20, size=100), 2), 0, 100),    # np.clip은 범위 밖으로 나가는 데이터를 잘라줌
    '영어' : np.clip(np.round(np.random.normal(70, 20, size=100), 2), 0, 100),
    '역사' : np.clip(np.round(np.random.normal(50, 20, size=100), 2), 0, 100),
}

df = pd.DataFrame(di)
df.to_csv("./data/scores.csv", index=False, encoding="utf-8-sig")
````

````python
# 필터링하기
cond = df['역사'] > 100.0
df[cond]
````

````python
# 국어 60점 이상, 영어 80점 이상인 학생들만 가져오기
cond = (df['국어'] >= 60.0) & (df['영어'] >= 80.0)
df[cond]
````

````python
# 데이터 추가하기
# 총점이라는 데이터를 생성하여 컬럼을 추가함
df['총점'] = df['국어'] + df['수학'] + df['영어'] + df['역사']
df['평균'] = (df['총점'] / 4).round(2)
df.head()
````

````python
# 100명 전체의 등수 구하기
df['석차'] = df['총점'].rank(ascending=False).astype(int)
df.to_csv("./data/scores.csv", index=False, encoding="utf-8-sig")
````

````python
# 데이터 정렬하기
df.sort_values(by='석차')
````

````python
df_agg = df.groupby('반')['총점'].aggregate(['count', 'mean', 'max', 'min', 'median']).round(2)
df_agg

# median: 오름차순으로 정렬한뒤 중앙값
# mode: 최빈값
````

````python
# 각 반별 평균 점수 표준편차 구하기

df_std = df.groupby(by='반')['평균'].std().round(2)
df_std
````

````python
# 국어, 영어, 수학, 총점, 평균
# 전체 데이터의 기본 통계정보
cols = subjects + ['총점', '평균']
df_aggs = df.groupby(by='반')[cols]
df_aggs.describe()
df[cols].describe().round(2)
````

````python
# 데이터를 구간화하기 : 수치형 데이터들을 특정 범위로 나누기

# 절대 범위로 나누기: A(90 ~ 100), B(80~89), C(70~79) 등급
# 상대 범위로 나누기: 일정한 비율로 밖에 나눌 수 없음, 상위, 중상, 중하, 하위

# 등급 부여하기(절대 평가)
df['절대범위'] = pd.cut(df['평균'],  # 구간화할 데이터 설정
       bins=[0, 40, 50, 60, 70, 80],   # 구간 설정
       labels=['F', 'D', 'C', 'B', 'A'],
       right=True
       )
df.head()
````

````python
# 상대 구간 나누기(상대 평가) AICE 시험에 나올 수 있음!!!!!
df['상대등급'] = pd.qcut(df['총점'], # 구간화할 데이터
        q=4,    # 몇개의 구간으로 나눌지 설정
        labels=['하위', '중하', '중상', '상위'],
        )
df.head()
````

````python
# '합격여부' 라는 필드를 생성하여 평균점수가 60점 이상이면 '합격' 미만이면 '불합격'ArithmeticError

df['합격여부'] = pd.cut(df['평균'], bins=[0, 60, 100], labels=['불합격', '합격'], right=True)
df.head()
````

````python
df['합격여부'] = np.where(df['평균'] >= 60, '합격', '불합격')
df.head()
````

````python
# 수학 점수가 75점이상 이고 평균 점수가 70 이상인 학생들만 필터링하기

math_cond = (df['수학'] >= 50) & (df['평균'] >= 50)
df_math_excel = df[math_cond].head()
df_math_excel
````

````python
# pivot table 만들기 -> 집계

df_pivot = pd.pivot_table(df, # 데이터 set을 지정
               index='반',
               columns='절대등급',
               values='번호',   # 집계할 데이터 (unique한 데이터)
               aggfunc='count',
               fill_value=0,    # 만들어지지 않는 데이터이면 0으로 채움, 기본 값은 Nan
               observed=False
               )

df_pivot
````
