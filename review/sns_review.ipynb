```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 200
```

```python
df = pd.read_csv('../AICE/data/flight.csv')
df = df.drop('Unnamed: 0', axis=1)
df.head(3)
```

```python
df.info()
```

```python
display(df.describe(include='all').T)
print(df['airline'].unique(), type(df['airline'].unique()))
```

```python
plt.figure(figsize=(4, 3))
sns.histplot(df, bins=50, x='duration', kde=True)
```

```python
plt.figure(figsize=(4, 3))
dense = plt.hist(df['duration'], bins=50, edgecolor='#000000', linewidth=0.5)
plt.title('Probability density of Duration')
plt.xlabel('hours')

plt.show()
```

```python
axes = df.hist(bins=50, color='Red', edgecolor='#123456', figsize=(8, 5))

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 6))

axes[0].hist(df['duration'], bins=50, color="#009E5A", edgecolor="#4F46C9")
axes[0].set_title('duration')
axes[0].set_xlabel('Hours')
axes[0].set_ylabel('Counts')

axes[1].hist(df['price'], bins=50, color="#15E0FF", edgecolor="#1B0832")
axes[1].set_title('Price')
axes[1].set_xlabel('Dollars')
axes[1].set_ylabel('Counts')

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 6))

sns.histplot(df['duration'], bins=50, kde=True, color='#347834', edgecolor='#785634', ax=axes[0])
axes[0].set_title('Duration')
axes[0].set_xlabel('Hours')
axes[0].set_ylabel('Counts')

sns.histplot(df['price'], bins=50, kde=True, color='#783467', edgecolor='#237845', ax=axes[1])
axes[1].set_title('Price')
axes[1].set_xlabel('Dollars')
axes[1].set_ylabel('Counts')

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 6))

sns.regplot(data=df, x='price', y='days_left', scatter_kws={'alpha':0.3, 's':15}, line_kws={'color':'red'}, ax=axes[0])
# sns.regplot(x=df['price'], y=df['days_left'], scatter_kws={'alpha':0.3, 's':15}, line_kws={'color':'red'}, ax=axes[0])

sns.scatterplot(data=df, x='duration', y='price', hue='days_left', ax=axes[1])

plt.show()
```

```python
cali_df = pd.read_csv('../AICE/data/california.csv')
cali_df.head(3)
```

```python
display(cali_df.info())
cali_df.describe(include='all')
cali_df['ocean_proximity'].value_counts()
```

```python
cols = cali_df.select_dtypes(include='number').columns
cols
```

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 6))

for i, col in enumerate(cols):
    sns.boxenplot(data=cali_df, y=col, ax=axes.flat[i])
    axes.flat[i].set_title(col)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 5))

for i, col in enumerate(cols):
    plt.subplot(2, 5, i+1)
    sns.boxplot(data=cali_df, y=cali_df[col])
    plt.title(col)

plt.show()
```
