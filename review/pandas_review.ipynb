# Series
- Series is a column
- Has index-value relationship

```python
di = {
    'one' : 1,
    'two' : 2,
    'three' : 3,
    'four' : 4,
    'five' : 5
}
s1 = pd.Series(di)
s1
```

```python
print(s1['one'])
print(s1.loc['one'])
print(s1.iloc[0])
```

```python
li = [1,2,3,4,5]
s1 = pd.Series(np.arange(1, 6))
label = ['one', 'two', 'three', 'four', 'five']

# You can't relabel an existing series like below
# s1 = pd.Series(s1, index=label)

# Instead, you can use .index
s1.index = label
s1
```

```python
# Slicing
print(s1['two':'four'])
print(s1['three':])
print(s1[::2])
print(s1[::-1])
```

```python
print(s1.values)
print()
print(s1.value_counts())
print()
print(s1.index)
print()
print(s1.array)
```

```python
s1['six'] = 6
s1['six-seven'] = 77
s1['six-seven'] = 67
s1
```

```python
s2 = s1.drop('six-seven')
print(s2)

s1.drop('six', inplace=True)
print(s1)
```

```python
s1['anomoly'] = 9867987
s1.drop(['anomoly', 'six-seven'], inplace=True)
s1
```

# DataFrame

- Tabular Data
- Rows and Columns
- A table

```python
# Ways to make dataframes
# 1. A dictionaary with a keyword and a list of a row
# 2. A list of dictionaries

df1 = pd.DataFrame({
    'name' : ['Kim', 'Lee', 'Choi'],
    'Age' : [23, 34, 13],
    'Gender' : ['Male', 'Female', 'Other']
})
print(df1)

df2 = pd.DataFrame([
    {'name' : 'Kim', 'Age' : 23, 'Gender' : 'Male'},
    {'name' : 'Lee', 'Age' : 34, 'Gender' : 'Female'},
    {'name' : 'Choi', 'Age' : 13, 'Gender' : 'Other'}
])
print(df2)
```
