```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../AICE/data/scores.csv')
df
```

```python
# groupby()
group_df = df.groupby(by='반')
subjects = ['국어', '영어', '수학', '역사']
display(group_df['역사'].agg(['mean', 'std', 'median']).round(2))
group_df[subjects].agg(['mean', 'std', 'median']).round(2)
```

```python
# 상대 구간 나누기, 절대 구간 나누기
df['전체석차'] = df['총점'].rank(ascending='False').astype(int)
df['학업성취도'] = pd.cut(df['전체석차'], bins=[0, 70, 80, 90, 100], right='True', labels=['하', '중하', '중상', '상'])
df['등급'] = pd.qcut(df['총점'], q=4, labels=['4등급', '3등급', '2등급', '1등급'])
df['합격여부'] = np.where(df['평균'] >= 60, '합격', '불합격')

df.head()
```
