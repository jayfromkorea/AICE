```python
import numpy as np
import matplotlib.pyplot as plt

apple = np.zeros((7, 7, 4), dtype=float)

# red apple body
apple[2:6, 1:6] = [1, 0, 0, 1]

# round it a tiny bit by clearing corners
apple[2, 1] = [0, 0, 0, 0]
apple[2, 5] = [0, 0, 0, 0]
apple[5, 1] = [0, 0, 0, 0]
apple[5, 5] = [0, 0, 0, 0]

# brown stem
apple[1, 3] = [0.45, 0.25, 0.05, 1]

# green leaf
apple[1, 4] = [0, 0.7, 0, 1]

plt.imshow(apple)
```
