```python
import pandas as pd
```

```python
# Series
li = [1,2,3,4,5]
s1 = pd.Series(li)
print(s1)
```

```python
# Labeling(indexing) a serires
lb = ['one', 'two', 'three', 'four', 'five']
# Need to give the list you'd like to make it a series, and the label list as the second argument with the keyword argument
s1 = pd.Series(li, index=lb)
print(s1['one'])
print(s1.loc['one'])
print(s1.iloc[0])
```
